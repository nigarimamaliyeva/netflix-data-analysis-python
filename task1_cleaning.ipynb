```python
df['director'] = df['director'].replace('Not Given', 'Unknown')
df['country'] = df['country'].replace('Not Given', 'Unknown')
```

```python
df.duplicated().sum()
```

```python
cols_to_check = df.columns.drop('show_id')
print(df.duplicated(subset=cols_to_check).sum())
```

```python
df[df.duplicated(subset=cols_to_check, keep=False)].sort_values('title')
```

```python
df = df.drop_duplicates(subset=cols_to_check, keep='first')
```

```python
print("Duplicates remaining:", df.duplicated(subset=cols_to_check).sum())
print("New shape:", df.shape)
```

```python
text_cols = ['show_id', 'type', 'title', 'director', 'country', 'rating', 'duration', 'listed_in']

for col in text_cols:
    print(col, (df[col] != df[col].str.strip()).sum())
```

```python
for col in text_cols:
    df[col] = df[col].str.strip()
```

```python
text_cols = ['show_id', 'type', 'title', 'director', 'country', 'rating', 'duration', 'listed_in']

for col in text_cols:
    print(col, (df[col] != df[col].str.strip()).sum())
```

```python
print("Unique countries:", df['country'].nunique())
print("Values changed by title-casing:", (df['country'] != df['country'].str.title()).sum())
```

```python
df['rating'] = df['rating'].replace({'NR': 'Not Rated', 'UR': 'Not Rated'})
```

```python
df['rating'].value_counts()
```

```python
print(df['date_added'].dtype)
df['date_added'].head(3)
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], format='%m/%d/%Y')
```

```python
print(df['date_added'].dtype)
print("Missing dates:", df['date_added'].isnull().sum())
print("Earliest:", df['date_added'].min())
print("Latest:", df['date_added'].max())
```

```python
df['duration'].value_counts().head(10)
```

```python
df['duration_value'] = df['duration'].str.extract(r'(\d+)')[0].astype(int)
df['duration_unit'] = (df['duration'].str.extract(r'\d+\s*(\D+)')[0]
                       .str.strip().replace({'Seasons': 'Season'}))
```

```python
print(df['duration_unit'].value_counts())
df.groupby('type')['duration_value'].describe()
```

```python
df = df.reset_index(drop=True)
```

## Task 1: Netflix Data Cleaning & Preparation - Summary

### Objective
Prepare the original Netflix dataset (`Dataset.csv`) for analysis and reporting.

### Steps Performed
1. **Import:** Loaded the dataset with pandas (8,790 rows, 10 columns).
2. **Missing values:** `isnull()` found no empty cells, but missing data was hidden as the text `'Not Given'` in `director` (2,588) and `country` (287). Replaced with `'Unknown'` so no rows were lost and no values were invented.
3. **Duplicates:** Compared all columns except `show_id` (which is unique per row) and found 3 records entered twice under different IDs. Kept the first copy of each, leaving 8,787 rows.
4. **Formatting:** Found and removed hidden trailing whitespace in one title (`title`). All other text columns were already clean.
5. **Standardization:**
   - `type`: already consistent (Movie / TV Show).
   - `country`: already consistent (86 unique values including Unknown).
   - `rating`: merged `NR` and `UR` into `Not Rated` (82 rows).
6. **Date format:** Converted `date_added` from text to datetime (range 2008-01-01 to 2021-09-25, no missing dates).
7. **Duration:** Split `duration` into numeric `duration_value` and `duration_unit` (`min` / `Season`), merging `Season` and `Seasons` into one label.
8. **Export:** Saved the result to `Dataset_cleaned.csv` (original file left unchanged).

### Final Dataset
- **Rows:** 8,787
- **Columns:** 12 (10 original + `duration_value`, `duration_unit`)
- **Movies:** 6,124 | **TV Shows:** 2,663
